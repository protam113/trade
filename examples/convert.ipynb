```python
import pandas as pd
import os

# File CSV
file_name = "trade_history_filtered.csv"
file_path = os.path.join(os.getcwd(), file_name)

if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print(f"✅ CSV loaded: {file_path}")

    if df.empty:
        print("⚠ CSV is empty.")
    else:
        # Tổng lệnh
        total_trades = len(df)

        # Lọc lệnh thắng / lệnh thua
        df_win = df[df['profit'] > 0].copy()
        df_loss = df[df['profit'] < 0].copy()

        num_win = len(df_win)
        num_loss = len(df_loss)

        # Tổng profit / tổng loss
        total_profit = df_win['profit'].sum()
        total_loss = df_loss['profit'].sum()  # âm

        # Winrate & Profit Factor
        winrate = num_win / total_trades * 100 if total_trades > 0 else 0
        profit_factor = total_profit / abs(total_loss) if total_loss != 0 else float('inf')

        # In thống kê
        print("=== Trade Performance Summary ===")
        print(f"Total trades: {total_trades}")
        print(f"Winning trades: {num_win}")
        print(f"Losing trades: {num_loss}")
        print(f"Total profit: {total_profit:.2f}")
        print(f"Total loss: {total_loss:.2f}")
        print(f"Winrate: {winrate:.2f}%")
        print(f"Profit factor: {profit_factor:.2f}")

else:
    print(f"⚠ CSV not found: {file_path}")
```
