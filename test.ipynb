```python
import os
import pandas as pd

# Tên file CSV gốc
file_name = "my_trades.csv"
file_path = os.path.join(os.getcwd(), file_name)

# Tên file CSV mới
sl_file_name = "sl.csv"
sl_file_path = os.path.join(os.getcwd(), sl_file_name)

if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print(f"✅ CSV exists: {file_path}")

    if df.empty:
        print("⚠ CSV is empty.")
    else:
        if 'comment' in df.columns:
            # Lọc các trade comment bắt đầu bằng "[sl"
            sl_trades = df[df['comment'].str.startswith('[sl', na=False)]
            print(f"Found {len(sl_trades)} trades with comment starting with '[sl'")

            if not sl_trades.empty:
                # Lưu ra file mới
                sl_trades.to_csv(sl_file_path, index=False)
                print(f"✅ Saved filtered trades to: {sl_file_path}")
            else:
                print("⚠ No trades with '[sl' comment found.")
        else:
            print("⚠ CSV không có cột 'comment'.")
else:
    print(f"⚠ CSV not found: {file_path}")
```
