```python
import pandas as pd

DATA_PATH = '../data/EURUSD_5M.csv'

# Đọc file gốc
df = pd.read_csv(DATA_PATH, parse_dates=['datetime'])
df.set_index('datetime', inplace=True)

# Hàm chuyển khung thời gian
def resample_data(df, timeframe, filename):
    rule = {
        '15M': '15T',
        '1H': '1H',
        '4H': '4H',
    }[timeframe]

    df_resampled = df.resample(rule).agg({
        'open': 'first',
        'high': 'max',
        'low': 'min',
        'close': 'last',
        'volume': 'sum'
    }).dropna()

    df_resampled.to_csv(filename)
    print(f"✅ Saved: {filename} ({len(df_resampled)} rows)")

# Tạo 3 file
resample_data(df, '15M', '../data/EURUSD_15M.csv')
resample_data(df, '1H', '../data/EURUSD_1H.csv')
resample_data(df, '4H', '../data/EURUSD_4H.csv')
```
